```python
from youtube_transcript_api import YouTubeTranscriptApi
```

```python
ytt_api = YouTubeTranscriptApi()

transcript = ytt_api.fetch('T-D1OfcDW1M')

print(transcript)
```

```python
transcript_text = " ".join(
    snippet.text for snippet in transcript
)

print(transcript_text[:2000])
```

```python
print("Number of transcript snippets:", len(transcript))
print("Total characters:", len(transcript_text))
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

chunks = text_splitter.split_text(transcript_text)

print("Number of chunks:", len(chunks))
```

```python
for i, chunk in enumerate(chunks[:3]):
    print(f"\n--- Chunk {i+1} ---")
    print(chunk)
```

```python
from sentence_transformers import SentenceTransformer
```

```python
embedding_model = SentenceTransformer("all-MiniLM-L6-v2")

print("Embedding model loaded successfully!")
```

```python
embeddings = embedding_model.encode(chunks)

print("Number of embeddings:", len(embeddings))
print("Embedding size:", embeddings.shape)  
```

```python
import faiss
embedding_dimension = embeddings.shape[1]

print("Embedding dimension:", embedding_dimension)
```

```python
index = faiss.IndexFlatL2(embedding_dimension)

print("FAISS index created!")
index.add(embeddings)

print("Number of vectors in FAISS:", index.ntotal)
```

```python
query = "What is this video about?"
query_embedding = embedding_model.encode([query])
distances, indices = index.search(query_embedding, k=3)

print("Distances:", distances)
print("Indices:", indices)
```

```python
for i in indices[0]:
    print("\n--- Retrieved Chunk ---")
    print(chunks[i])
```

```python
import ollama

response = ollama.chat(
    model="qwen3:4b",
    messages=[
        {
            "role": "user",
            "content": "What is Retrieval Augmented Generation? Explain simply."
        }
    ]
)

print(response["message"]["content"])
```

```python
def retrieve_context(query, k=3):
    
    query_embedding = embedding_model.encode([query])
    
    distances, indices = index.search(query_embedding, k)
    
    retrieved_chunks = []
    
    for i in indices[0]:
        retrieved_chunks.append(chunks[i])
    
    return retrieved_chunks
```

```python
query = "What does retrieval augmented generation mean?"

retrieved_chunks = retrieve_context(query)

for i, chunk in enumerate(retrieved_chunks):
    print(f"\n--- Chunk {i+1} ---")
    print(chunk)
```

```python
context = "\n\n".join(retrieved_chunks)

prompt = f"""
Answer the user's question using only the information provided in the context.

If the answer is not available in the context, say:
"I could not find this information in the YouTube video."

Context:
{context}

User Question:
{query}
"""
```

```python
def check_youtube_relevance(query, threshold=1.5):
    
    query_embedding = embedding_model.encode([query])
    
    distances, indices = index.search(query_embedding, k=3)
    
    best_distance = distances[0][0]
    
    if best_distance <= threshold:
        return True, best_distance
    else:
        return False, best_distance
```

```python
query = "What is retrieval augmented generation?"

is_relevant, distance = check_youtube_relevance(query)

print("Distance:", distance)
print("YouTube relevant:", is_relevant)
```

```python
query = "What is the capital of France?"

is_relevant, distance = check_youtube_relevance(query)

print("Distance:", distance)
print("YouTube relevant:", is_relevant)
```

```python
def check_context_with_llm(query, retrieved_chunks):
    
    context = "\n\n".join(retrieved_chunks)

    prompt = f"""
You are checking whether the provided context can answer the question.

Context:
{context}

Question:
{query}

Can the context answer the question?

Reply with only:
YES
or
NO
"""

    response = ollama.chat(
        model="qwen3:4b",
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ]
    )

    result = response["message"]["content"].strip().upper()

    if "YES" in result:
        return True
    else:
        return False
```

```python
query = "What is retrieval augmented generation?"

retrieved_chunks = retrieve_context(query)

result = check_context_with_llm(query, retrieved_chunks)

print("Can YouTube answer?", result)
```

```python
def web_search(query, max_results=3):
    results = tavily_client.search(
        query=query,
        max_results=max_results
    )
    
    web_results = []
    
    for result in results["results"]:
        web_results.append({
            "title": result["title"],
            "url": result["url"],
            "content": result["content"]
        })
    
    return web_results
```

```python
def route_query(query):
    
    # Retrieve YouTube context
    retrieved_chunks = retrieve_context(query)
    
    # Check whether YouTube is relevant
    is_relevant = check_context_with_llm(
        query,
        retrieved_chunks
    )
    
    if is_relevant:
        return "youtube", retrieved_chunks
    
    else:
        web_results = web_search(query)
        return "web", web_results
```

```python
from typing import TypedDict

class AgentState(TypedDict):
    query: str
    source: str
    context: list
    answer: str
```

```python
def youtube_node(state):
    
    query = state["query"]
    
    # Step 1: similarity check
    is_similar, distance = check_youtube_relevance(query)
    
    if not is_similar:
        return {
            "source": "web",
            "context": []
        }
    
    # Step 2: LLM verification
    retrieved_chunks = retrieve_context(query)
    
    is_relevant = check_context_with_llm(
        query,
        retrieved_chunks
    )
    
    if is_relevant:
        return {
            "source": "youtube",
            "context": retrieved_chunks
        }
    
    else:
        return {
            "source": "web",
            "context": []
        }
```

```python
state = {
    "query": "What is Retrieval Augmented Generation?",
    "source": "",
    "context": [],
    "answer": ""
}

result = youtube_node(state)

print(result)
```

```python
def web_node(state):
    
    query = state["query"]
    
    web_results = web_search(query)
    
    return {
        "source": "web",
        "context": web_results
    }
```

```python
state = {
    "query": "What is the capital of France?",
    "source": "",
    "context": [],
    "answer": ""
}

result = web_node(state)

print(result["source"])
print(result["context"])
```

```python
def route_node(state):
    
    if state["source"] == "youtube":
        return "youtube"
    
    else:
        return "web"
```

```python
from langgraph.graph import StateGraph, END
```

```python
graph = StateGraph(AgentState)

graph.add_node("youtube", youtube_node)
graph.add_node("web", web_node)

graph.set_entry_point("youtube")

graph.add_conditional_edges(
    "youtube",
    route_node,
    {
        "youtube": END,
        "web": "web"
    }
)

graph.add_edge("web", END)
```

```python
agent = graph.compile()

print("Agent created successfully!")
```

```python
state = {
    "query": "What is Retrieval Augmented Generation?",
    "source": "",
    "context": [],
    "answer": ""
}

result = agent.invoke(state)

print("Source:", result["source"])
print("Context:")
print(result["context"])
```

```python
def answer_node(state):
    
    query = state["query"]
    context = state["context"]
    source = state["source"]
    
    if source == "youtube":
        context_text = "\n\n".join(context)
        
    else:
        context_text = "\n\n".join(
            [item["content"] for item in context]
        )
    
    prompt = f"""
Answer the user's question using only the provided context.

Context:
{context_text}

Question:
{query}

Give a clear and simple answer.
"""
    
    response = ollama.chat(
        model="qwen3:4b",
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ]
    )
    
    return {
        "answer": response["message"]["content"]
    }
```

```python
graph.add_node("answer", answer_node)
```

```python
graph = StateGraph(AgentState)

graph.add_node("youtube", youtube_node)
graph.add_node("web", web_node)
graph.add_node("answer", answer_node)

graph.set_entry_point("youtube")

graph.add_conditional_edges(
    "youtube",
    route_node,
    {
        "youtube": "answer",
        "web": "web"
    }
)

graph.add_edge("web", "answer")
graph.add_edge("answer", END)

agent = graph.compile()

print("Final agent created!")
```

```python
state = {
    "query": "What is Retrieval Augmented Generation?",
    "source": "",
    "context": [],
    "answer": ""
}

result = agent.invoke(state)

print("Source:", result["source"])
print("\nFinal Answer:")
print(result["answer"])
```
